# House Prediction

##### Downloading the data


```python
import pandas as pd
import numpy as np
df = pd.read_csv("data/USA Housing Dataset.csv")
df.head()
```

```python
other_input = ['bedrooms', 'bathrooms', 'floors', 'condition']
to_scale = ['sqft_living','sqft_lot','sqft_above','yr_built']
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler().fit(df[to_scale])
scaled_input = scaler.transform(df[to_scale])
scaled_input
```

```python
all = np.concatenate((scaled_input, df[other_input].values), axis = 1)
model = LinearRegression().fit(all, df['price'])
model
```

```python
prediction = model.predict([[0.56,0.34,0.45,0.65,1,1,1,3]])
prediction
```

```python
prediction = model.predict(df[['bedrooms', 'bathrooms', 'floors', 'condition','sqft_living','sqft_lot','sqft_above','yr_built']])
prediction
```

```python

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
mean_absolute_error(df['price'], prediction)
```
