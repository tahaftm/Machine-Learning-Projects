# House Prediction

##### Downloading the data


```python
import pandas as pd
import numpy as np
df = pd.read_csv("data/USA Housing Dataset.csv")
df.head()
```

```python
other_input = ['bedrooms', 'bathrooms', 'floors', 'condition']
to_scale = ['sqft_living','sqft_lot','sqft_above',"yr_built"]
```

### Building the model

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler().fit(df[to_scale])
scaled_input = scaler.transform(df[to_scale])
df[to_scale] = scaled_input
df
```

```python
# all = np.concatenate((df[to_scale].values, df[other_input].values), axis = 1)
all = pd.concat([df[to_scale], df[other_input]], axis=1)

from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(all, df['price'], test_size=0.3, random_state=42)

model = LinearRegression().fit(X_train, Y_train)
```

### Testing the model

```python
prediction = model.predict(pd.DataFrame([[0.2, -0.4, 0.5, -0.9, 2,3,1,3]], columns=all.columns))
print(f"The house price is : {prediction[0]}")
```

### Evaluating the model

```python
prediction = model.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
print(f"Mean absolute error: {mean_absolute_error(Y_test, prediction)}")
print(f"Mean squared error: {mean_squared_error(Y_test, prediction)}")
print(f"r2 score: {r2_score(Y_test, prediction)}")
```
