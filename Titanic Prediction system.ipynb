# Titanic Prediction System

### Downloading the dataset

```python
import pandas as pd 
df = pd.read_csv("data/Titanic-Dataset.csv", index_col=0)
df.dropna(subset=['Survived'], inplace=True)
target_ser = df['Survived']
df
```

### Data Preprocessing

Defining inputs and target columns 

```python
target_cols = 'Survived'
input_cols = df.drop(columns=[target_cols,'Name']).columns.tolist() 
```

Defining the numerical and categorical columns

```python
numerical_cols = df.drop(columns= target_cols).select_dtypes(include=['int64', 'float64']).columns.tolist()
categorical_cols = df.drop(columns=[target_cols,'Name']).select_dtypes(include=['object']).columns.tolist()
```

Imputing missing values

```python
df.isna().sum()
```

```python
from sklearn.impute import SimpleImputer
num_imputer = SimpleImputer(strategy='mean')
num_imputer.fit(df[numerical_cols])
df[numerical_cols] = num_imputer.transform(df[numerical_cols])
cat_imputer = SimpleImputer(strategy='most_frequent')
cat_imputer.fit(df[categorical_cols])
df[categorical_cols] = cat_imputer.transform(df[categorical_cols])
df.isna().sum()
```

Scaling the Data

```python
df.describe()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler().fit(df[numerical_cols])
df[numerical_cols] = scaler.transform(df[numerical_cols])
```

Encoding the data

```python
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
encoded_cat = encoder.fit_transform(df[categorical_cols])
encoded_cols = encoder.get_feature_names_out(categorical_cols)
encoded_df = pd.DataFrame(encoded_cat, columns=encoded_cols, index=df.index)
data = pd.concat([df,encoded_df], axis = 1).drop(columns=target_cols)
pd.set_option('display.max_columns', None)
data
```

```python
training_data = pd.concat([data[numerical_cols], data[encoded_cols]], axis=1)
training_data
```

Splitting the data 

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(training_data, target_ser, test_size=0.2, random_state=42)
```

### Building and Evaluating the model

```python
from sklearn.ensemble import RandomForestClassifier
model = RandomForestClassifier(n_estimators=500,n_jobs=-1,max_depth=300, random_state=42,class_weight='balanced',max_features=5)
model.fit(X_train, y_train)
xTrain_pred = model.predict(X_train)
xTrain_pred
```

```python
testing_pred = model.predict(X_test)
testing_pred
```

```python
from sklearn.metrics import accuracy_score
print(f"The accuracy achieved by Decision Tree: {accuracy_score(testing_pred ,y_test)*100:.2f}%")
```

```python
from sklearn.linear_model import LogisticRegression
model2 = LogisticRegression(solver = 'liblinear')
model2.fit(X_train, y_train)
testing_pred = model2.predict(X_test)
testing_pred
```

```python
print(f"The accuracy achieved by Logistic Regression: {accuracy_score(testing_pred ,y_test)*100:.2f}%")
```

### Creating a function to predict

```python
def predict_input(input_dict):
    input_df = pd.DataFrame([input_dict])
    input_df[categorical_cols] = cat_imputer.transform(input_df[categorical_cols], )
    input_df[numerical_cols] = num_imputer.transform(input_df[numerical_cols])
    input_df[numerical_cols] = scaler.transform(input_df[numerical_cols])
    encoded_df = pd.DataFrame(encoder.transform(input_df[categorical_cols]), columns=encoded_cols)
    input_df = pd.concat([input_df,encoded_df],axis=1)
    X_input = pd.concat([input_df[numerical_cols], encoded_df],axis = 1)
    pred = model.predict(X_input)
    if pred == 0:
        print("They are dead!")
    else:
        print("They are survivors!")
```

```python
predict_input({
        "PassengerId": 892,
        "Survived": 0,
        "Pclass": 3,
        "Name": "Kelly, Mr. James",
        "Sex": "male",
        "Age": 34.5,
        "SibSp": 0,
        "Parch": 0,
        "Ticket": "PC 17599",
        "Fare": 7.8292,
        "Cabin": None,
        "Embarked": "Q"
    })
```

### Saving and loading the model

```python
titanic_model = {
    'Logistic_model' : model2,
    'Forest_model' : model,
    'numerical_imputer' : num_imputer,
    'categorical_imputer' : cat_imputer,
    'scaler' : scaler,
    'evaluate' : predict_input
}
```

```python
import joblib
joblib.dump(titanic_model, 'joblibs models/titanic_model.joblib')
```
